```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/BigBasket Products.csv')
print("Dataset loaded successfully!")
```

```python
# Step 2: Preview the first 12 rows
print("--- Step 2: Top 12 Rows ---")
display(df.head(12))

# Step 3: Summary statistics of numeric columns
print("\n--- Step 3: Data Description ---")
display(df.describe())

# Step 4: Dataset structure, data types, and non-null counts
print("\n--- Step 4: Data Info ---")
df.info()
```

```python
# Step 5: Top 10 and Bottom 10 Most Frequent Products in the Catalog
print("--- Top 10 Most Frequent Products ---")
display(df["product"].value_counts().head(10))

print("\n--- Bottom 10 Least Frequent Products ---")
display(df["product"].value_counts().tail(10))

# Step 6: Calculate Discount and Discount Percentage
df["discount"] = df["market_price"] - df["sale_price"]
df["discount_percent"] = np.where(
    (df["market_price"].notna()) & (df["market_price"] > 0),
    ((df["discount"] / df["market_price"]) * 100).round(2),
    np.nan,
)

# Display discount details for the first product
print("\n--- Step 6: Discount on First Item in Dataset ---")
display(
    df[
        [
            "product",
            "brand",
            "market_price",
            "sale_price",
            "discount",
            "discount_percent",
        ]
    ].head(1)
)

# Display discount details for a specific product
target_product = "Garlic Oil - Vegetarian Capsule 500 mg"
print(f"\n--- Discount on '{target_product}' ---")
display(
    df[df["product"] == target_product][
        ["product", "market_price", "sale_price", "discount", "discount_percent"]
    ]
)
```

```python
# Step 7: Check missing values in each column
print("--- Step 7: Missing Values in Each Column ---")
missing_values = df.isnull().sum()
display(missing_values[missing_values > 0])

# Step 8: Detect outliers using the IQR method
numeric_cols = ["sale_price", "market_price", "rating"]

print("\n--- Step 8: Outlier Detection (IQR Method) ---")
for col in numeric_cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_limit = q1 - 1.5 * iqr
  upper_limit = q3 + 1.5 * iqr

  outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
  print(f"Total outliers found in '{col}': {len(outliers)}")

  # Optional: Cap extreme values within IQR bounds while preserving original column
  df[f"{col}_capped"] = df[col].clip(lower=lower_limit, upper=upper_limit)

print("\nSummary statistics of numeric columns:")
display(df[numeric_cols].describe())
```

```python
# Step 9: Create Visualizations
fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# Graph 1: Top 10 Categories
top_categories = df["category"].value_counts().head(10)
sns.barplot(
    x=top_categories.values,
    y=top_categories.index,
    hue=top_categories.index,
    palette="viridis",
    legend=False,
    ax=axes[0, 0],
)
axes[0, 0].set_title(
    "Top 10 Product Categories on BigBasket", fontsize=12, fontweight="bold"
)
axes[0, 0].set_xlabel("Number of Products")
axes[0, 0].set_ylabel("Category")

# Graph 2: Top 10 Brands
top_brands = df["brand"].value_counts().head(10)
sns.barplot(
    x=top_brands.values,
    y=top_brands.index,
    hue=top_brands.index,
    palette="magma",
    legend=False,
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    "Top 10 Brands on BigBasket", fontsize=12, fontweight="bold"
)
axes[0, 1].set_xlabel("Number of Products")
axes[0, 1].set_ylabel("Brand")

# Graph 3: Distribution of Ratings
sns.histplot(
    df["rating"].dropna(), bins=20, kde=True, color="teal", ax=axes[1, 0]
)
axes[1, 0].set_title(
    "Product Ratings Distribution", fontsize=12, fontweight="bold"
)
axes[1, 0].set_xlabel("Rating (Out of 5)")
axes[1, 0].set_ylabel("Count")

# Graph 4: Market Price vs Sale Price
sns.scatterplot(
    data=df,
    x="market_price",
    y="sale_price",
    alpha=0.5,
    color="coral",
    ax=axes[1, 1],
)
axes[1, 1].set_title(
    "Market Price vs Sale Price", fontsize=12, fontweight="bold"
)
axes[1, 1].set_xlabel("Market Price (INR)")
axes[1, 1].set_ylabel("Sale Price (INR)")

plt.tight_layout()
plt.savefig("bigbasket_eda_overview.png", dpi=300, bbox_inches="tight")
plt.show()
```
